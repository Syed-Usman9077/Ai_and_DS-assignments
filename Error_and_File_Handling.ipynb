### Q4 
Ek file `numbers.txt` banayein aur us mein 1 se 5 tak numbers line-by-line write karein.

```python
file = open("numbers.txt", "w")
for i in range(1, 6):
    file.write(str(i) + "\n")
file.close()
```

### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
file = open("numbers.txt", "r")
total = 0
for number in file:
    total = total + int(number)
print(total)
file.close()
```

### Q7 (Medium)
`with open()` statement ka istemal karte hue ek file `data.txt` read karein aur explain karein ki `with` ka kya faida hai.

```python
with open("data.txt", "w") as file:
    file.write("Python File Handling")

with open("data.txt", "r") as file:
    print(file.read())

print("with open() automatically closes the file after use.")
```

### Q9 
Ek file `source.txt` ka data read karke `destination.txt` file mein copy karein.

```python
with open("source.txt", "w") as file:
    file.write("This is source file data.")

with open("source.txt", "r") as source:
    data = source.read()

with open("destination.txt", "w") as destination:
    destination.write(data)
```

### Q10

### Q11

### Q12 (Easy)

### Q13 (Easy)
Aisi file open karne ki koshish karein jo exist nahi karti (`non_existent_file.txt`). `FileNotFoundError` ko `try-except` se handle karein.

```python
try:
    file = open("non_existent_file.txt", "r")
    print(file.read())
    file.close()
except FileNotFoundError:
    print("File not found")
```

### Q14 
Ek program banayein jisme `finally` block ho, jo humesha execute ho chahe error aaye ya na aaye.

```python
try:
    number = int(input("Enter a number: "))
    print(10 / number)
except (ValueError, ZeroDivisionError):
    print("An error occurred")
finally:
    print("Finally block executed")
```

### Q15 
Ek list `my_list = [10, 20, 30]` banayein. Invalid index (e.g., index 5) access karein aur `IndexError` catch karein.

### Q16

### Q17
Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print(result)
except ZeroDivisionError:
    print("Cannot divide by zero")
except ValueError:
    print("Please enter a valid number")
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Invalid input")
else:
    print("You entered:", number)
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    print("Age is valid")

try:
    age = int(input("Enter your age: "))
    check_age(age)
except ValueError as error:
    print(error)
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
file = None

try:
    file = open("destination.txt", "r")
    print(file.read())
except FileNotFoundError:
    print("File not found")
finally:
    if file is not None:
        file.close()
    print("File handling completed")
```
